# 最小二乘法求一元线性回归
y = 4x+2
## 构造数据集

```python
import numpy as np
np.random.seed(0)
X=np.random.normal(size=(100,1),scale=1)
# X
```

```python
y=4*X[:,0]+2
# y
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
X.shape,y.shape
```

```python
all_date=np.concatenate((X,y.reshape(100,1)),axis=1)
all_date.shape
```

```python
np.random.shuffle(all_date)
```
